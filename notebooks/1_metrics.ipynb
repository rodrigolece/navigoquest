```python
import pathlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from navigoquest.utils.config import standard_dirs, LEVELS, METADATA_COLS, AGE_RANGE
from navigoquest import format_metric_task_data
from navigoquest.utils.pipelines import load_environment, load_paths_normative, load_paths_clinical, pipeline_compute_features
```

```python
# Settings common over levels
early_stop = np.inf
use_sparse_norm = True
output_filenames = {lvl: f"metrics_level{lvl:02}_gb_2480.csv" for lvl in np.arange(1, 99)}
task_list = ['voc', 'path_length', 'average_curvature', 'boundary_affinity',
             'frobenius_deviation', 'supremum_deviation', 'conformity', 'vector_conformity']
levels_now = LEVELS

# Directories
dirs = standard_dirs()
paths_dir = dirs['paths']
output_dir = dirs['output']
if not output_dir.exists():
    output_dir.mkdir(parents=True)
env_store = load_environment(dirs)
```

# 1. Benchmark dataset

```python
# Run and save feature computation
output_df_all = {}
# for lvl in LEVELS:
for lvl in levels_now:
    # Arguments for the pipeline
    path_data = load_paths_normative(lvl, dirs, AGE_RANGE)
    env_data = env_store.get(lvl)
    task_data = format_metric_task_data(task_list)

    # Main computation
    pipeline_output = pipeline_compute_features(path_data, env_data, task_data, 
                                                early_stop, use_sparse_norm)

    # Format output
    output_df = pd.DataFrame.from_records(pipeline_output)
    output_df.to_csv(output_dir / output_filenames[lvl], index=False)
    output_df_all[lvl] = output_df
```

# 2. Clinical dataset

```python
filename_save = "clinical_metrics.csv"

output_all = []
for lvl in LEVELS:
    # Arguments for the pipeline
    path_data = load_paths_clinical(lvl, dirs)
    env_data = env_store.get(lvl)
    task_data = format_metric_task_data(task_list)

    # Main computation
    output = pipeline_compute_features(path_data, env_data, task_data, np.inf, use_sparse_norm)
    output_all.extend(output)

# Collate output and save
output_df = pd.DataFrame.from_records(output_all)
output_df.to_csv(output_dir / filename_save, index=False)
```
